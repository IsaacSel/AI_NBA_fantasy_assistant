```python

import pandas as pd

path = "/kaggle/input/datasets/eoinamoore/historical-nba-data-and-player-box-scores/Players.csv"

players = pd.read_csv(path, low_memory=False)

# Check the actual columns
print(players.columns.tolist())

# Identify NBA position flags
flags = {"guard": "G", "forward": "F", "center": "C"}

for col in flags:
    players[col] = pd.to_numeric(
        players[col], errors="coerce"
    ).fillna(0).astype(int)

# Combine positions, allowing G/F and F/C
players["position"] = players.apply(
    lambda row: "/".join(
        label for col, label in flags.items()
        if row[col] == 1
    ) or "Unknown",
    axis=1
)

positions = players[
    ["personId", "firstName", "lastName", "position"]
]

positions.to_csv(
    "/kaggle/working/nba_player_positions.csv",
    index=False
)

print(positions.head(10))
```
